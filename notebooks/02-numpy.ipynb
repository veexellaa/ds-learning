```python
import numpy as np

m = np.arange(1, 21).reshape(4, 5)
print(m)
print(m.shape)

print(m[:, 2])   # третий столбец
print(m[3])      
# последняя строка
print(m % 3)

mask = m % 3 == 0
print(mask)

print(m[mask])
print(m[mask].sum())
```

```python
m2 = m.copy()
m2[m>15] = -1
print(m2)
```
